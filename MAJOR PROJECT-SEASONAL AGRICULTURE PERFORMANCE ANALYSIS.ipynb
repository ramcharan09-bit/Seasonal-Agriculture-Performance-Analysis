# **7.Data Cleaning**

Data cleaning is an important step in preparing the dataset for analysis
In ths step, the data datset is checked for missing values and duplicate records

```python
# Check for duplicate rows
duplicate_count=df.duplicated().sum()
print("Number of duplicate rows",duplicate_count)
```

```python
# Remove duplicate rows and reset index
df=df.drop_duplicates().reset_index(drop=True)
print("Shape after removing duplicates:",df.shape)
```

```python
# Display value counts for 'Season' column
df['Season'].value_counts()
```

```python
# Calculate average production by 'Season'
df.groupby('Season')['Production_Tonnes'].mean()
```

```python
# Calculate average production, yield, and rainfall by 'Season'
df.groupby('Season')[['Production_Tonnes','Yield_Tonnes_Ha','Rainfall_mm']].mean()
```

```python
# Display data types of each column
df.dtypes
```

```python
# Generate descriptive statistics for object columns
df.describe(include="object")
```

```python
# Summarize missing values by count and percentage
missing=df.isnull().sum().sort_values(ascending=False)
missing_percentage=(df.isnull().mean()*100).sort_values(ascending=False)
missing_summary=pd.DataFrame({"Missing_Count":missing,"missing_percentage":missing_percentage})
missing_summary[missing_summary["Missing_Count"]>0]
```

```python
# Visualize missing values
plt.figure(figsize=(10,6))
missing=df.isnull().sum().sort_values(ascending=False)
missing_plot=missing[missing>0]
if len(missing_plot)>0:
  sns.barplot(x=missing_plot.index,y=missing_plot.values)
  plt.xticks(rotation=60,ha="right")
  plt.ylabel("Number of missing values")
  plt.xlabel("column")
  plt.title("Missing values in each column")
  plt.show()
```

```python
# Explore categorical columns
categorical_columns=df.select_dtypes(include="object").columns
for col in categorical_columns:
  print("columns:")
  print(df[col].nunique(),"unique values")
  print(df[col].dropna().unique()[:30])
```

```python
# Display top 15 columns with missing values
df.isnull().sum().sort_values(ascending=False).head(15)
```

```python
# Impute missing numerical values with the median
numeric_columns=df.select_dtypes(include=np.number).columns
for col in numeric_columns:
  if df[col].isnull().sum()>0:
    df[col]=df[col].fillna(df[col].median())
print("Remainging Missing Values:",df.isnull().sum().sum())
```

```python
# Separate numerical and categorical columns
numeric_columns=df.select_dtypes(include=np.number).columns.tolist()
categorical_columns=df.select_dtypes(include="object").columns.tolist()
print("Numerical Columns:")
print(numeric_columns)
print("Categorical Columns:")
print(categorical_columns)
```

## **9.Data Visualization**

We used bar charts, histograms, Pie chart, scatter plots, box plots and heatmaps to analyze and visualize the agricultural data



```python
# Create a bar plot of production by season
sns.barplot(data=df,x='Season',y='Production_Tonnes')
plt.title('Production by season')
plt.show()
```

```python
plt.figure(figsize=(7,6))
sns.countplot(data=df,x="Season")
plt.title("Number of Agriculture Records per Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
```

```python
avg_yield_by_season = df.groupby('Season')['Yield_Tonnes_Ha'].mean().reset_index()
plt.figure(figsize=(7, 5))
sns.barplot(x='Season', y='Yield_Tonnes_Ha', data=avg_yield_by_season, palette='viridis', hue='Season', legend=False)
plt.title('Average Yield (Tonnes/Ha) by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.show()
```

```python
avg_profit_by_season = df.groupby('Season')['Profit_INR'].mean().reset_index()
plt.figure(figsize=(8, 6))
sns.barplot(x='Season', y='Profit_INR', data=avg_profit_by_season, palette='viridis', hue='Season', legend=False)
plt.title('Average Profit (INR) by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.show()
```

```python
farms_by_season = df['Season'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(farms_by_season, labels=farms_by_season.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('viridis'))
plt.title('Number of Farms by Season')
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Yield_Tonnes_Ha'], bins=30, kde=True)
plt.title('Distribution of Crop Yield (Tonnes/Ha)')
plt.xlabel('Yield (Tonnes/Ha)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Season', y='Yield_Tonnes_Ha', data=df, palette='viridis', hue='Season', legend=False)
plt.title('Distribution of Yield (Tonnes/Ha) by Season')
plt.xlabel('Season')
plt.ylabel('Yield (Tonnes/Ha)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Rainfall_mm', y='Yield_Tonnes_Ha', data=df)
plt.title('Rainfall vs Yield')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Water_Used_m3', y='Yield_Tonnes_Ha', data=df)
plt.title('Water Usage vs Yield')
plt.xlabel('Water Used (m³)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Fertilizer_kg_ha', y='Yield_Tonnes_Ha', data=df)
plt.title('Fertilizer Usage vs Yield')
plt.xlabel('Fertilizer (kg/ha)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(16, 12))
sns.heatmap(df.select_dtypes(include=np.number).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

```python
production_by_season = df.groupby('Season')['Production_Tonnes'].sum().reset_index()

plt.figure(figsize=(10, 6))
plt.stackplot(production_by_season['Season'], production_by_season['Production_Tonnes'], labels=production_by_season['Season'])
plt.title('Production by Season (Area Chart)')
plt.xlabel('Season')
plt.ylabel('Total Production (Tonnes)')
plt.legend(loc='upper left')
plt.grid(True)
plt.show()
```

```python
avg_profit_by_season = df.groupby('Season')['Profit_INR'].mean().reset_index()

plt.figure(figsize=(8, 6))
sns.lineplot(x='Season', y='Profit_INR', data=avg_profit_by_season, marker='o')
plt.title('Average Profit (INR) by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Season', y='Profit_INR', data=df, palette='viridis', hue='Season', legend=False)
plt.title('Distribution of Profit (INR) by Season')
plt.xlabel('Season')
plt.ylabel('Profit (INR)')
plt.show()
```

```python
irrigation_methods = df['Irrigation_Method'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(irrigation_methods, labels=irrigation_methods.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Distribution of Irrigation Methods')
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Irrigation_Method', y='Water_Used_m3', data=df, palette='pastel', hue='Irrigation_Method', legend=False)
plt.title('Distribution of Water Used (m³) by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Water Used (m³)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Disease_Pest_Risk_pct'], bins=20, kde=True)
plt.title('Distribution of Disease and Pest Risk Percentage')
plt.xlabel('Disease and Pest Risk (%)')
plt.ylabel('Frequency')
plt.show()
```

```python
season_crop = df.pivot_table(values="Yield_Tonnes_Ha",index="Crop",columns="Season",aggfunc="mean")
plt.figure(figsize=(10,7))
sns.heatmap(season_crop,annot=True,fmt=".2f")
plt.title("Average Crop Yield by Season")
plt.xlabel("Season")
plt.ylabel("Crop")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Total_Cost_INR"],df["Revenue_INR"],alpha=0.3)
plt.title("Revenue vs Total Cost")
plt.xlabel("Total Cost (INR)")
plt.ylabel("Revenue (INR)")
plt.show()
```

Based on the analysis, the following recommendations can be considered:

2. Consider seasonal rainfall and temperature patterns when selecting crops and planning cultivation

3. Improve water management by identifying seasons and farming practices with better water-use efficiency

4. Monitor disease and pest risk more closely during seasons with higher observed risk
